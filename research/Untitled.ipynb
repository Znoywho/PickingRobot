```python
import numpy as np


# INITILIZE
def initilizeGridMap(w, h, n_r):

    GRID_W = w
    GRID_H = h
    NUMBER_OF_RACKS = n_r

    GRID_MAP = np.zeros((GRID_W, GRID_H), dtype=int)

    occupied = set()
    rack_positions = []

    while len(rack_positions) < NUMBER_OF_RACKS:
        x, y = np.random.randint(0, GRID_W - 1), np.random.randint(0, GRID_H - 1)
        if (x, y) not in occupied:
            occupied.add((x, y))
            rack_positions.append((x, y))

            GRID_MAP[x][y] = 1
    return GRID_MAP, occupied, rack_positions
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Create a 3D grid structure
# We define a 1x1x1 spatial grid array filled with True values
filled = np.ones((1, 1, 1), dtype=bool)

# 2. Initialize the plotting window with 3D projection
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# 3. Draw the 3D block
# Use 'alpha' to adjust transparency and 'edgecolor' to define borders
ax.voxels(filled, facecolors='teal', edgecolors='black', alpha=0.8)

# 4. Set clear spatial labels
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')
ax.set_title('3D Block Visualization')

# 5. Fix axis proportions to keep the block shape cubic
ax.set_aspect('equal')

# Display the render
```
